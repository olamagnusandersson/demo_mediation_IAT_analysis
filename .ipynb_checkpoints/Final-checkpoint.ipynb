```python
# Fixed causal data generator (fully-consistent generative logit, no calibration)
# Paste & run this in your Python environment.

import numpy as np
import pandas as pd
import scipy.special as sp
import os

np.random.seed(42)

# --- CONFIG ---
N = 10000
OUT_DIR = './causal_datasets_consistent'  # change path as needed
os.makedirs(OUT_DIR, exist_ok=True)

# Outcome logit components (change these if you want to tune marginal churn)
OUT_INTERCEPT = 1.0
BETA_INCOME = -0.0002
BETA_PROMO = 0.4

# Two parameterizations illustrating the two causal stories
PARAMS_A = {
    'label': 'Dataset_A_SelfSelection_NDE_gt_NIE',
    'latent_w': [0.5, 0.5],
    'latent_on_m': 1,   # explicit latent -> mediator
    't_on_m': 1.0,
    'm_on_y': -1.0,
    't_on_y': -2.0,
    'latent_on_y': -5.0
}
PARAMS_B = {
    'label': 'Dataset_B_ProductEffect_NIE_gt_NDE',
    'latent_w': [0.5, 0.5],
    'latent_on_m': 1.,
    't_on_m': 1.0,
    'm_on_y': -1.0,
    't_on_y': -0.2,
    'latent_on_y': -1.5
}

def generate_confouders(n):
    """Generate Income and Promotions confounders."""
    inc = np.random.lognormal(10.5, 0.5, n)
    inc_scaled = (inc - inc.mean()) / inc.std()
    promo = np.random.binomial(1, sp.expit(inc_scaled), n)
    return inc, inc_scaled, promo

def generate_data(params, income, income_scaled, promotions):
    """Generate dataset using the canonical structural model (no post-hoc calibration)."""
    n = len(income)
    # Latent loyalty (unobserved confounder)
    noise_L = np.random.normal(0, 1, n)
    latent_L = params['latent_w'][0] * income_scaled + params['latent_w'][1] * promotions + noise_L

    # Treatment assignment (Num_Products)
    p2 = sp.expit(0.8 * latent_L - 0.5)
    p3 = sp.expit(0.8 * latent_L - 1.5)
    num_prods = 1 + np.random.binomial(1, p2, size=n) + np.random.binomial(1, p3, size=n)

    # Mediator (IAT) with its own noise
    noise_M = np.random.normal(0, 1, n)
    m = params['latent_on_m'] * latent_L + params['t_on_m'] * num_prods + noise_M

    # Outcome probability comes from the same logit used in get_ground_truth
    logit = (
        OUT_INTERCEPT +
        BETA_INCOME * income +
        BETA_PROMO * promotions +
        params['latent_on_y'] * latent_L +
        params['t_on_y'] * num_prods +
        params['m_on_y'] * m
    )
    churn_prob = sp.expit(logit)
    churn = np.random.binomial(1, churn_prob, size=n)

    df = pd.DataFrame({
        'Income': income,
        'Income_Scaled': income_scaled,
        'Promotions': promotions,
        'Num_Products': num_prods,
        'Loyalty_IAT_Score': m,
        'Churn_Prob': churn_prob,
        'Churn': churn,
        'Source': params['label'],
        '_Latent_L': latent_L,
        '_Noise_M': noise_M
    })
    return df

def get_ground_truth(df, params, t0=1, t1=3):
    """Compute TE, NDE, NIE using stored latent and mediator noise (true ground-truth)."""
    latent_L = df['_Latent_L'].values
    noise_M = df['_Noise_M'].values

    def mediator_cf(t):
        return params['latent_on_m'] * latent_L + params['t_on_m'] * t + noise_M

    def outcome_prob_cf(t, m):
        logit = (
            OUT_INTERCEPT +
            BETA_INCOME * df['Income'].values +
            BETA_PROMO * df['Promotions'].values +
            params['latent_on_y'] * latent_L +
            params['t_on_y'] * t +
            params['m_on_y'] * m
        )
        return sp.expit(logit)

    m1 = mediator_cf(t1)
    m0 = mediator_cf(t0)
    te = (outcome_prob_cf(t1, m1) - outcome_prob_cf(t0, m0)).mean()
    nde = (outcome_prob_cf(t1, m0) - outcome_prob_cf(t0, m0)).mean()
    nie = te - nde
    return te, nde, nie

# --- Generate data (confounders shared across datasets) ---
inc_raw, inc_std, prm_raw = generate_confouders(N)

df_a = generate_data(PARAMS_A, inc_raw, inc_std, prm_raw)
df_b = generate_data(PARAMS_B, inc_raw, inc_std, prm_raw)

# --- Sanity checks (you can expand or automate these) ---
print("Sanity check dataset sizes:", len(df_a), len(df_b))
print("Num_Products distribution A:", df_a['Num_Products'].value_counts(normalize=True).sort_index().to_dict())
print("Num_Products distribution B:", df_b['Num_Products'].value_counts(normalize=True).sort_index().to_dict())
print("Overall churn A:", df_a['Churn'].mean())
print("Overall churn B:", df_b['Churn'].mean())
print("Churn by products A:", df_a.groupby('Num_Products')['Churn'].mean().to_dict())
print("Churn by products B:", df_b.groupby('Num_Products')['Churn'].mean().to_dict())

# Compute ground-truth causal effects
gt_a = get_ground_truth(df_a, PARAMS_A)
gt_b = get_ground_truth(df_b, PARAMS_B)
print("Ground truth (TE, NDE, NIE) for A:", gt_a)
print("Ground truth (TE, NDE, NIE) for B:", gt_b)

# Save datasets
df_a.to_csv(os.path.join(OUT_DIR, 'dataset_a_consistent.csv'), index=False)
df_b.to_csv(os.path.join(OUT_DIR, 'dataset_b_consistent.csv'), index=False)
print("Datasets saved to:", OUT_DIR)
```

```python
# Causal data generator with intercept-based calibration (preserves logit causality)
# Run locally. Prints TE, NDE, NIE and overall churn for both datasets.

import numpy as np
import pandas as pd
import scipy.special as sp
import os
import json

np.random.seed(42)

# --- CONFIG ---
N = 10000
OUT_DIR = './calibrated_datasets'
os.makedirs(OUT_DIR, exist_ok=True)

# Desired per-product churn targets (surface story)
TARGETS = {1: 0.20, 2: 0.10, 3: 0.05}
TARGETS_A = {1: 0.2, 2: 0.15, 3: 0.1}   # Strong product story
TARGETS_B = {1: 0.2, 2: 0.15, 3: 0.1}   # Weaker product story


# Base outcome logit coefficients (you can tune OUT_INTERCEPT if needed)
OUT_INTERCEPT = 0.0
BETA_INCOME = -0.0002
BETA_PROMO = 0.4

# Two parameter sets (A: self-selection / NDE > NIE, B: product-effect / NIE > NDE)

PARAMS_A = {
    'label': 'Dataset_A',
    'latent_w': [.1, .1],
    'latent_on_m': 0.1,
    't_on_m': 0.1,
    'm_on_y': -0.1,
    't_on_y': -0.3,
    'latent_on_y': -0.5
}

PARAMS_B = {
    'label': 'Dataset_B',
    'latent_w': [0.1, 0.1],
    'latent_on_m': 0.1,
    't_on_m': 0.9,
    'm_on_y': -0.9,
    't_on_y': -0.1,
    'latent_on_y': -0.5
}

# --- Utilities ---
def generate_confounders(n):
    inc = np.random.lognormal(10.5, 0.5, n)
    inc_scaled = (inc - inc.mean()) / inc.std()
    promo = np.random.binomial(1, sp.expit(inc_scaled), n)
    return inc, inc_scaled, promo

def solve_intercept_for_target(logit_vals, target, low=-40, high=40, tol=1e-9, maxiter=80):
    """Find a scalar a such that mean(sigmoid(logit_vals + a)) = target via bisection."""
    def f(a): return sp.expit(logit_vals + a).mean() - target
    lo, hi = low, high
    flo, fhi = f(lo), f(hi)
    # Expand if no sign change
    expand = 0
    while flo * fhi > 0 and expand < 6:
        lo *= 2; hi *= 2
        flo, fhi = f(lo), f(hi)
        expand += 1
    # Bisection
    for _ in range(maxiter):
        mid = 0.5 * (lo + hi)
        fmid = f(mid)
        if abs(fmid) < tol:
            return mid
        if flo * fmid <= 0:
            hi = mid; fhi = fmid
        else:
            lo = mid; flo = fmid
    return mid

def generate_and_calibrate(params, income, income_scaled, promotions, targets):
    n = len(income)
    # Latent
    noise_L = np.random.normal(0, 1, n)
    latent_L = params['latent_w'][0]*income_scaled + params['latent_w'][1]*promotions + noise_L

    # Treatment assignment
    p2 = sp.expit(0.8*latent_L - 0.5)
    p3 = sp.expit(0.8*latent_L - 1.5)
    num_prods = 1 + np.random.binomial(1, p2, size=n) + np.random.binomial(1, p3, size=n)

    # Mediator noise and precomputed mediator under each t
    noise_M = np.random.normal(0, 1, n)
    mediator_t = {t: params['latent_on_m'] * latent_L + params['t_on_m'] * t + noise_M for t in [1,2,3]}

    # Base logits (no per-treatment shift)
    base_logit_t = {}
    for t in [1,2,3]:
        base_logit_t[t] = (OUT_INTERCEPT + BETA_INCOME * income + BETA_PROMO * promotions +
                           params['latent_on_y'] * latent_L + params['t_on_y'] * t + params['m_on_y'] * mediator_t[t])

    # Solve for treatment-specific intercepts so E[Y|T=t] = target_t

    overall_target = np.mean([targets[t] for t in num_prods])

    # Stack logits under observed treatment
    observed_logits = np.empty(n)
    for t in [1,2,3]:
        mask = (num_prods == t)
        observed_logits[mask] = base_logit_t[t][mask]

    shared_intercept = solve_intercept_for_target(observed_logits, overall_target)

    treatment_intercepts = {1: shared_intercept, 2: shared_intercept, 3: shared_intercept}

    # Compute final churn probability using intercept for each unit's observed T
    churn_prob = np.empty(n)
    for t in [1,2,3]:
        mask = (num_prods == t)
        if mask.any():
            churn_prob[mask] = sp.expit(base_logit_t[t][mask] + treatment_intercepts[t])

    churn = np.random.binomial(1, churn_prob, size=n)

    # Observed mediator: the mediator under the observed T (for clarity we'll store actual observed m_obs)
    m_obs = np.empty(n)
    for t in [1,2,3]:
        mask = (num_prods == t)
        m_obs[mask] = mediator_t[t][mask]

    df = pd.DataFrame({
        'Income': income,
        'Income_Scaled': income_scaled,
        'Promotions': promotions,
        'Num_Products': num_prods,
        'Loyalty_IAT_Score': m_obs,
        'Churn_Prob': churn_prob,
        'Churn': churn,
        'Source': params['label'],
        '_Latent_L': latent_L,
        '_Noise_M': noise_M
    })

    # Attach extras for ground-truth calculation
    df.attrs['treatment_intercepts'] = treatment_intercepts
    df.attrs['mediator_t'] = mediator_t
    df.attrs['base_logit_t'] = base_logit_t
    return df

def get_ground_truth(df, params, t0=1, t1=3):
    latent_L = df['_Latent_L'].values
    noise_M = df['_Noise_M'].values
    intercepts = df.attrs.get('treatment_intercepts', {1:0,2:0,3:0})
    def mediator_cf(t):
        return params['latent_on_m'] * latent_L + params['t_on_m'] * t + noise_M
    def outcome_prob_cf(t, m):
        a_t = intercepts[t]
        logit = (OUT_INTERCEPT + BETA_INCOME * df['Income'].values + BETA_PROMO * df['Promotions'].values +
                 params['latent_on_y'] * latent_L + params['t_on_y'] * t + params['m_on_y'] * m + a_t)
        return sp.expit(logit)
    m1 = mediator_cf(t1); m0 = mediator_cf(t0)
    te = (outcome_prob_cf(t1, m1) - outcome_prob_cf(t0, m0)).mean()
    nde = (outcome_prob_cf(t1, m0) - outcome_prob_cf(t0, m0)).mean()
    nie = te - nde
    return te, nde, nie

# --- Run generation ---
inc_raw, inc_std, prm_raw = generate_confounders(N)
df_a = generate_and_calibrate(PARAMS_A, inc_raw, inc_std, prm_raw, TARGETS_A)
df_b = generate_and_calibrate(PARAMS_B, inc_raw, inc_std, prm_raw, TARGETS_B)

# Compute ground truth and print concise summary
gt_a = get_ground_truth(df_a, PARAMS_A)
gt_b = get_ground_truth(df_b, PARAMS_B)

def print_summary(label, gt, df):
    te, nde, nie = gt
    overall = df['Churn'].mean()
    print("="*60)
    print(f"{label}")
    print(f"Total Effect (TE)   : {te: .6f}")
    print(f"Natural Direct (NDE): {nde: .6f}")
    print(f"Natural Indirect(NIE): {nie: .6f}")
    print(f"Overall Churn Rate   : {overall:.4%}")
    print("="*60 + "\n")

print_summary(PARAMS_A['label'], gt_a, df_a)
print_summary(PARAMS_B['label'], gt_b, df_b)

# Save outputs (datasets + intercepts)
df_a.to_csv(os.path.join(OUT_DIR, 'dataset_a_calibrated.csv'), index=False)
df_b.to_csv(os.path.join(OUT_DIR, 'dataset_b_calibrated.csv'), index=False)
with open(os.path.join(OUT_DIR, 'intercepts_a.json'), 'w') as f:
    json.dump(df_a.attrs['treatment_intercepts'], f)
with open(os.path.join(OUT_DIR, 'intercepts_b.json'), 'w') as f:
    json.dump(df_b.attrs['treatment_intercepts'], f)

print("Saved datasets to:", OUT_DIR)
# Combine datasets (assumes df_a and df_b already exist)
combined = pd.concat([df_a, df_b], ignore_index=True)

sns.set_theme(style="whitegrid")

plt.figure(figsize=(8, 5))
sns.barplot(
    data=combined,
    x='Num_Products',
    y='Churn',
    hue='Source',
    errorbar=None
)

plt.title("Observed Churn Rate by Number of Products", fontweight='bold')
plt.ylabel("Churn Rate")
plt.xlabel("Number of Products")
plt.gca().yaxis.set_major_formatter(
    plt.FuncFormatter(lambda y, _: f"{y:.0%}")
)

plt.tight_layout()
plt.show()
```

```python
# === Mediation estimation using 1m approach (regressor + probability clipping) ===

from sklearn.model_selection import KFold
from lightgbm import LGBMRegressor
import numpy as np
from econml.dml import LinearDML

def run_mediation_dml(
    data,
    confounders,
    t0=1,
    t1=3,
    T_col='Num_Products',
    n_splits=5,
    random_state=42
):
    """
    Binary-treatment mediation (Option 1m):
      T = 1{Num_Products == t1} vs 0{Num_Products == t0}

    Estimates:
      - TE via LinearDML (diagnostic)
      - NDE / NIE via cross-fitted sequential g-computation (regressor)
    """

    # ----------------------------
    # 0) Build binary treatment
    # ----------------------------
    T_raw = data[T_col].to_numpy()
    T = (T_raw == t1).astype(float)   # binary: 1 if t1, 0 if t0

    Y = data['Churn'].to_numpy(dtype=float)
    M = data['Loyalty_IAT_Score'].to_numpy(dtype=float)
    X = data[confounders].to_numpy(dtype=float)

    n = len(Y)

    # ----------------------------
    # 1) TE via LinearDML (binary)
    # ----------------------------
    model_y = LGBMRegressor(n_estimators=200, random_state=random_state)
    model_t = LGBMRegressor(n_estimators=200, random_state=random_state + 1)

    dml_te = LinearDML(
        model_y=model_y,
        model_t=model_t,
        discrete_treatment=True,
        cv=n_splits,
        random_state=random_state
    )
    dml_te.fit(Y, T, X=X)
    te_dml = float(dml_te.effect(X).mean())

    # ----------------------------
    # 2) Cross-fitted E[M | X, T]
    # ----------------------------
    kf = KFold(n_splits=n_splits, shuffle=True, random_state=random_state)

    m_hat_0 = np.zeros(n)  # M(t0)
    m_hat_1 = np.zeros(n)  # M(t1)

    for tr, te in kf.split(X):
        reg_m = LGBMRegressor(n_estimators=200, random_state=random_state)
        reg_m.fit(
            np.hstack([X[tr], T[tr].reshape(-1, 1)]),
            M[tr]
        )
        m_hat_0[te] = reg_m.predict(np.hstack([X[te], np.zeros((len(te), 1))]))
        m_hat_1[te] = reg_m.predict(np.hstack([X[te], np.ones((len(te), 1))]))

    # ----------------------------
    # 3) Cross-fitted E[Y | X, T, M] (regressor)
    # ----------------------------
    y_1_m1 = np.zeros(n)
    y_1_m0 = np.zeros(n)
    y_0_m0 = np.zeros(n)

    for tr, te in kf.split(X):
        reg_y = LGBMRegressor(n_estimators=300, random_state=random_state + 10)
        reg_y.fit(
            np.hstack([X[tr], T[tr].reshape(-1,1), M[tr].reshape(-1,1)]),
            Y[tr]
        )
        X_te = X[te]

        y_1_m1[te] = reg_y.predict(np.hstack([X_te, np.ones((len(te),1)), m_hat_1[te].reshape(-1,1)]))
        y_1_m0[te] = reg_y.predict(np.hstack([X_te, np.ones((len(te),1)), m_hat_0[te].reshape(-1,1)]))
        y_0_m0[te] = reg_y.predict(np.hstack([X_te, np.zeros((len(te),1)), m_hat_0[te].reshape(-1,1)]))

    # Probability clipping
    y_1_m1 = np.clip(y_1_m1, 0, 1)
    y_1_m0 = np.clip(y_1_m0, 0, 1)
    y_0_m0 = np.clip(y_0_m0, 0, 1)

    # ----------------------------
    # 4) Effects
    # ----------------------------
    te_gc  = float(np.mean(y_1_m1 - y_0_m0))
    nde_gc = float(np.mean(y_1_m0 - y_0_m0))
    nie_gc = te_gc - nde_gc

    delta_m = float(np.mean(m_hat_1 - m_hat_0))

    return {
        'te_gc': te_gc,
        'nde_gc': nde_gc,
        'nie_gc': nie_gc,
        'te_dml_diag': te_dml,
        'mean_delta_m': delta_m,
        'overall_churn': float(Y.mean())
    }



# --- 5. EXECUTION & ANALYSIS (Unchanged) ---
conf_cols = ['Income', 'Promotions']

# Map dataset name to the actual DataFrame and the structural parameters
datasets_to_analyze = [
    ("Dataset A (Self-Selection)", df_a, PARAMS_A),
    ("Dataset B (Product Effect)", df_b, PARAMS_B)
]

for name, df, params in datasets_to_analyze:
    print(f"\n{'='*60}\nANALYZING: {name}\n{'='*60}")
    
    # 1. Ground Truth (true causal effects from the DGP)
    gt_te, gt_nde, gt_nie = get_ground_truth(df, params)
    
    # 2. Mediation estimation (DML + cross-fitted g-computation)
    results = run_mediation_dml(df, conf_cols)

    est_te  = results['te_gc']
    est_nde = results['nde_gc']
    est_nie = results['nie_gc']

    # Diagnostics (always present in current implementation)
    te_dml_diag   = results['te_dml_diag']
    mean_delta_m  = results['mean_delta_m']
    overall_churn = results['overall_churn']

    # 3. Print results (concise and robust)
    print(f"Ground Truth: TE={gt_te:.3f}, NDE={gt_nde:.3f}, NIE={gt_nie:.3f}")
    print(f"Estimated  : TE={est_te:.3f}, NDE={est_nde:.3f}, NIE={est_nie:.3f}")
    print(
        f"(Diagnostic) LinearDML TE = {te_dml_diag:.3f} | "
        f"mean ΔM (t1−t0) = {mean_delta_m:.4f} | "
        f"obs churn = {overall_churn:.4%}"
    )
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import KFold
from lightgbm import LGBMRegressor, LGBMClassifier
from econml.dml import LinearDML

# --- Global experiment log ---
EXPERIMENT_LOG = []

def run_mediation_experiment(
    data,
    confounders,
    t0=1,
    t1=3,
    T_col='Num_Products',
    n_splits=5,
    random_state=42,
    use_classifier=True,
    clip_probs=True,
    experiment_label="default"
):
    """
    Experimental DML + cross-fitted g-computation mediation.
    Logs each experiment for analysis.
    """
    # --- Binary treatment ---
    T_raw = data[T_col].to_numpy()
    T = (T_raw == t1).astype(float)

    Y = data['Churn'].to_numpy(dtype=float)
    M = data['Loyalty_IAT_Score'].to_numpy(dtype=float)
    X = data[confounders].to_numpy(dtype=float)
    n = len(Y)

    # --- LinearDML TE (diagnostic) ---
    model_y = LGBMRegressor(n_estimators=200, random_state=random_state)
    model_t = LGBMRegressor(n_estimators=200, random_state=random_state+1)
    dml_te = LinearDML(model_y=model_y, model_t=model_t, discrete_treatment=True,
                       cv=n_splits, random_state=random_state)
    dml_te.fit(Y, T, X=X)
    te_dml = float(dml_te.effect(X).mean())

    # --- Cross-fitted mediator ---
    kf = KFold(n_splits=n_splits, shuffle=True, random_state=random_state)
    m_hat_0 = np.zeros(n)
    m_hat_1 = np.zeros(n)
    for tr, te in kf.split(X):
        reg_m = LGBMRegressor(n_estimators=200, random_state=random_state)
        reg_m.fit(np.hstack([X[tr], T[tr].reshape(-1,1)]), M[tr])
        m_hat_0[te] = reg_m.predict(np.hstack([X[te], np.zeros((len(te),1))]))
        m_hat_1[te] = reg_m.predict(np.hstack([X[te], np.ones((len(te),1))]))

    # --- Cross-fitted outcome ---
    y_1_m1 = np.zeros(n)
    y_1_m0 = np.zeros(n)
    y_0_m0 = np.zeros(n)
    for tr, te in kf.split(X):
        if use_classifier:
            reg_y = LGBMClassifier(n_estimators=300, random_state=random_state+10)
            reg_y.fit(np.hstack([X[tr], T[tr].reshape(-1,1), M[tr].reshape(-1,1)]), Y[tr])
            y_1_m1[te] = reg_y.predict_proba(np.hstack([X[te], np.ones((len(te),1)), m_hat_1[te].reshape(-1,1)]))[:,1]
            y_1_m0[te] = reg_y.predict_proba(np.hstack([X[te], np.ones((len(te),1)), m_hat_0[te].reshape(-1,1)]))[:,1]
            y_0_m0[te] = reg_y.predict_proba(np.hstack([X[te], np.zeros((len(te),1)), m_hat_0[te].reshape(-1,1)]))[:,1]
        else:
            reg_y = LGBMRegressor(n_estimators=300, random_state=random_state+10)
            reg_y.fit(np.hstack([X[tr], T[tr].reshape(-1,1), M[tr].reshape(-1,1)]), Y[tr])
            y_1_m1[te] = reg_y.predict(np.hstack([X[te], np.ones((len(te),1)), m_hat_1[te].reshape(-1,1)]))
            y_1_m0[te] = reg_y.predict(np.hstack([X[te], np.ones((len(te),1)), m_hat_0[te].reshape(-1,1)]))
            y_0_m0[te] = reg_y.predict(np.hstack([X[te], np.zeros((len(te),1)), m_hat_0[te].reshape(-1,1)]))

    if clip_probs:
        y_1_m1 = np.clip(y_1_m1, 0, 1)
        y_1_m0 = np.clip(y_1_m0, 0, 1)
        y_0_m0 = np.clip(y_0_m0, 0, 1)

    # --- Compute effects ---
    te_gc = float(np.mean(y_1_m1 - y_0_m0))
    nde_gc = float(np.mean(y_1_m0 - y_0_m0))
    nie_gc = te_gc - nde_gc
    delta_m = float(np.mean(m_hat_1 - m_hat_0))
    overall_churn = float(Y.mean())

    # --- Log experiment ---
    EXPERIMENT_LOG.append({
        "label": experiment_label,
        "te_gc": te_gc,
        "nde_gc": nde_gc,
        "nie_gc": nie_gc,
        "te_dml": te_dml,
        "mean_delta_m": delta_m,
        "overall_churn": overall_churn,
        "use_classifier": use_classifier,
        "clip_probs": clip_probs
    })

    return {
        'te_gc': te_gc,
        'nde_gc': nde_gc,
        'nie_gc': nie_gc,
        'te_dml_diag': te_dml,
        'mean_delta_m': delta_m,
        'overall_churn': overall_churn
    }

# --- Run multiple experiments ---
experiment_settings = [
    {"use_classifier": True,  "clip_probs": True,  "label": "cls_clip"},
    {"use_classifier": True,  "clip_probs": False, "label": "cls_no_clip"},
    {"use_classifier": False, "clip_probs": True,  "label": "reg_clip"},
    {"use_classifier": False, "clip_probs": False, "label": "reg_no_clip"}
]

for name, df, params in datasets_to_analyze:
    print(f"\n{'='*60}\nANALYZING: {name}\n{'='*60}")
    gt_te, gt_nde, gt_nie = get_ground_truth(df, params)
    print(f"Ground Truth: TE={gt_te:.3f}, NDE={gt_nde:.3f}, NIE={gt_nie:.3f}\n")

    for exp in experiment_settings:
        res = run_mediation_experiment(df, conf_cols,
                                       use_classifier=exp['use_classifier'],
                                       clip_probs=exp['clip_probs'],
                                       experiment_label=exp['label'])
        print(f"[{exp['label']}] Estimated: TE={res['te_gc']:.3f}, NDE={res['nde_gc']:.3f}, NIE={res['nie_gc']:.3f} | "
              f"ΔM={res['mean_delta_m']:.2f}, LinearDML TE={res['te_dml_diag']:.3f}")
```

```python
from sklearn.model_selection import KFold
from lightgbm import LGBMRegressor, LGBMClassifier
from sklearn.linear_model import LinearRegression
import numpy as np
from econml.dml import LinearDML

def run_mediation_dml_experiments(
    data,
    confounders,
    t0=1,
    t1=3,
    T_col='Num_Products',
    n_splits=5,
    random_state=42
):
    """
    Run multiple mediation estimation strategies (baseline + experiments)
    Returns a dict keyed by experiment name with TE/NDE/NIE and diagnostics
    """
    results_dict = {}
    
    # Data arrays
    T_raw = data[T_col].to_numpy()
    Y = data['Churn'].to_numpy(dtype=float)
    M = data['Loyalty_IAT_Score'].to_numpy(dtype=float)
    X = data[confounders].to_numpy(dtype=float)
    n = len(Y)

    # ----------------------------
    # Binary treatment for baseline and experiments
    # ----------------------------
    T_bin = (T_raw == t1).astype(float)

    # Experiment configurations
    experiments = {
        'reg_clip':  {'M_model': LGBMRegressor, 'Y_model': LGBMRegressor, 'clip': True},
        'cls_clip':  {'M_model': LGBMRegressor, 'Y_model': LGBMClassifier, 'clip': True},
        '1m':        {'M_model': LinearRegression, 'Y_model': LGBMRegressor, 'clip': True},
        '2':         {'M_model': LGBMRegressor, 'Y_model': LinearRegression, 'clip': True},
        '3_contT':   {'M_model': LGBMRegressor, 'Y_model': LGBMRegressor, 'clip': True, 'continuous_T': True}
    }

    for exp_name, cfg in experiments.items():
        # ----------------------------
        # Treatment
        # ----------------------------
        if cfg.get('continuous_T', False):
            T = T_raw.astype(float)
        else:
            T = T_bin

        # ----------------------------
        # 1) TE via LinearDML
        # ----------------------------
        model_y = LGBMRegressor(n_estimators=200, random_state=random_state)
        model_t = LGBMRegressor(n_estimators=200, random_state=random_state + 1)

        dml_te = LinearDML(
            model_y=model_y,
            model_t=model_t,
            discrete_treatment=not cfg.get('continuous_T', False),
            cv=n_splits,
            random_state=random_state
        )
        dml_te.fit(Y, T, X=X)
        te_dml = float(dml_te.effect(X).mean())

        # ----------------------------
        # 2) Cross-fitted E[M | X, T]
        # ----------------------------
        kf = KFold(n_splits=n_splits, shuffle=True, random_state=random_state)
        m_hat_0 = np.zeros(n)
        m_hat_1 = np.zeros(n)

        for tr, te_idx in kf.split(X):
            reg_m = cfg['M_model']()
            reg_m.fit(np.hstack([X[tr], T[tr].reshape(-1,1)]), M[tr])
            m_hat_0[te_idx] = reg_m.predict(np.hstack([X[te_idx], np.zeros((len(te_idx),1))]))
            m_hat_1[te_idx] = reg_m.predict(np.hstack([X[te_idx], np.ones((len(te_idx),1))]))

        # ----------------------------
        # 3) Cross-fitted E[Y | X, T, M]
        # ----------------------------
        y_1_m1 = np.zeros(n)
        y_1_m0 = np.zeros(n)
        y_0_m0 = np.zeros(n)

        for tr, te_idx in kf.split(X):
            reg_y = cfg['Y_model']()
            reg_y.fit(np.hstack([X[tr], T[tr].reshape(-1,1), M[tr].reshape(-1,1)]), Y[tr])
            X_te = X[te_idx]

            if isinstance(reg_y, LGBMClassifier):
                y_1_m1[te_idx] = reg_y.predict_proba(np.hstack([X_te, np.ones((len(te_idx),1)), m_hat_1[te_idx].reshape(-1,1)]))[:,1]
                y_1_m0[te_idx] = reg_y.predict_proba(np.hstack([X_te, np.ones((len(te_idx),1)), m_hat_0[te_idx].reshape(-1,1)]))[:,1]
                y_0_m0[te_idx] = reg_y.predict_proba(np.hstack([X_te, np.zeros((len(te_idx),1)), m_hat_0[te_idx].reshape(-1,1)]))[:,1]
            else:
                y_1_m1[te_idx] = reg_y.predict(np.hstack([X_te, np.ones((len(te_idx),1)), m_hat_1[te_idx].reshape(-1,1)]))
                y_1_m0[te_idx] = reg_y.predict(np.hstack([X_te, np.ones((len(te_idx),1)), m_hat_0[te_idx].reshape(-1,1)]))
                y_0_m0[te_idx] = reg_y.predict(np.hstack([X_te, np.zeros((len(te_idx),1)), m_hat_0[te_idx].reshape(-1,1)]))

        # Optional clipping
        if cfg.get('clip', False):
            y_1_m1 = np.clip(y_1_m1, 0, 1)
            y_1_m0 = np.clip(y_1_m0, 0, 1)
            y_0_m0 = np.clip(y_0_m0, 0, 1)

        # ----------------------------
        # 4) Effects
        # ----------------------------
        te_gc = float(np.mean(y_1_m1 - y_0_m0))
        nde_gc = float(np.mean(y_1_m0 - y_0_m0))
        nie_gc = te_gc - nde_gc
        delta_m = float(np.mean(m_hat_1 - m_hat_0))

        results_dict[exp_name] = {
            'TE': te_gc,
            'NDE': nde_gc,
            'NIE': nie_gc,
            'LinearDML_TE': te_dml,
            'Delta_M': delta_m,
            'Churn': float(Y.mean())
        }

    return results_dict

# ----------------------------
# Run experiments
# ----------------------------
conf_cols = ['Income', 'Promotions']
datasets_to_analyze = [
    ("Dataset A (Self-Selection)", df_a, PARAMS_A),
    ("Dataset B (Product Effect)", df_b, PARAMS_B)
]

for name, df, params in datasets_to_analyze:
    print(f"\n{'='*60}\nANALYZING: {name}\n{'='*60}")
    gt_te, gt_nde, gt_nie = get_ground_truth(df, params)

    exp_results = run_mediation_dml_experiments(df, conf_cols)

    print(f"Ground Truth: TE={gt_te:.3f}, NDE={gt_nde:.3f}, NIE={gt_nie:.3f}\n")
    for exp_name, res in exp_results.items():
        print(f"[{exp_name}] Estimated: TE={res['TE']:.3f}, NDE={res['NDE']:.3f}, NIE={res['NIE']:.3f} | "
              f"ΔM={res['Delta_M']:.2f}, LinearDML TE={res['LinearDML_TE']:.3f}, Churn={res['Churn']:.4%}")
```

```python
# Paste this into your working notebook and run.
from sklearn.model_selection import KFold
from lightgbm import LGBMRegressor
from sklearn.linear_model import LinearRegression
import numpy as np
import pandas as pd
import os

# (If df_a/df_b already exist in your notebook, this will use them. Otherwise load saved CSVs)
try:
    df_a
    df_b
except NameError:
    # adjust paths if your CSVs are in another folder
    df_a = pd.read_csv('./calibrated_datasets/dataset_a_calibrated.csv')
    df_b = pd.read_csv('./calibrated_datasets/dataset_b_calibrated.csv')

conf_cols = ['Income', 'Promotions']

# Ensure types
for df in (df_a, df_b):
    df['Num_Products'] = df['Num_Products'].astype(int)
    df['Promotions'] = df['Promotions'].astype(int)
    df['Income'] = pd.to_numeric(df['Income'], errors='coerce').fillna(df['Income'].mean())

# The core experiment function (baseline + scaled + residual + continuous-T)
def run_experiments_on_df(df, confounders, t0=1, t1=3, n_splits=5, random_state=42):
    from sklearn.model_selection import KFold
    from lightgbm import LGBMRegressor
    from sklearn.linear_model import LinearRegression
    import numpy as np

    T_raw = df['Num_Products'].to_numpy()
    T_bin = (T_raw == t1).astype(float)
    Y = df['Churn'].to_numpy(dtype=float)
    M = df['Loyalty_IAT_Score'].to_numpy(dtype=float)
    X = df[confounders].to_numpy(dtype=float)
    n = len(Y)
    kf = KFold(n_splits=n_splits, shuffle=True, random_state=random_state)

    def baseline_reg_clip():
        m_hat_0 = np.zeros(n); m_hat_1 = np.zeros(n)
        for tr, te in kf.split(X):
            reg_m = LGBMRegressor(n_estimators=200, random_state=random_state)
            reg_m.fit(np.hstack([X[tr], T_bin[tr].reshape(-1,1)]), M[tr])
            m_hat_0[te] = reg_m.predict(np.hstack([X[te], np.zeros((len(te),1))]))
            m_hat_1[te] = reg_m.predict(np.hstack([X[te], np.ones((len(te),1))]))
        y_1_m1 = np.zeros(n); y_1_m0 = np.zeros(n); y_0_m0 = np.zeros(n)
        for tr, te in kf.split(X):
            reg_y = LGBMRegressor(n_estimators=300, random_state=random_state+10)
            reg_y.fit(np.hstack([X[tr], T_bin[tr].reshape(-1,1), M[tr].reshape(-1,1)]), Y[tr])
            X_te = X[te]
            y_1_m1[te] = reg_y.predict(np.hstack([X_te, np.ones((len(te),1)), m_hat_1[te].reshape(-1,1)]))
            y_1_m0[te] = reg_y.predict(np.hstack([X_te, np.ones((len(te),1)), m_hat_0[te].reshape(-1,1)]))
            y_0_m0[te] = reg_y.predict(np.hstack([X_te, np.zeros((len(te),1)), m_hat_0[te].reshape(-1,1)]))
        y_1_m1[:] = np.clip(y_1_m1, 0, 1); y_1_m0[:] = np.clip(y_1_m0, 0, 1); y_0_m0[:] = np.clip(y_0_m0, 0, 1)
        te = float(np.mean(y_1_m1 - y_0_m0)); nde = float(np.mean(y_1_m0 - y_0_m0)); nie = te-nde
        delta_m = float(np.mean(m_hat_1 - m_hat_0))
        return {'TE':te,'NDE':nde,'NIE':nie,'Delta_M':delta_m}

    def exp_scaled():
        m_hat_0 = np.zeros(n); m_hat_1 = np.zeros(n)
        for tr, te in kf.split(X):
            reg_m = LGBMRegressor(n_estimators=200, random_state=random_state)
            reg_m.fit(np.hstack([X[tr], T_bin[tr].reshape(-1,1)]), M[tr])
            m_hat_0[te] = reg_m.predict(np.hstack([X[te], np.zeros((len(te),1))]))
            m_hat_1[te] = reg_m.predict(np.hstack([X[te], np.ones((len(te),1))]))
        obs_mean0 = M[T_bin==0].mean() if (T_bin==0).sum()>0 else M.mean()
        obs_mean1 = M[T_bin==1].mean() if (T_bin==1).sum()>0 else M.mean()
        hat_mean0 = m_hat_0.mean(); hat_mean1 = m_hat_1.mean()
        delta_obs = obs_mean1 - obs_mean0
        delta_hat = hat_mean1 - hat_mean0
        s = (delta_obs / delta_hat) if abs(delta_hat) > 1e-6 else 0.0
        m_hat_0 = hat_mean0 + s*(m_hat_0 - hat_mean0)
        m_hat_1 = hat_mean1 + s*(m_hat_1 - hat_mean1)
        y_1_m1 = np.zeros(n); y_1_m0 = np.zeros(n); y_0_m0 = np.zeros(n)
        for tr, te in kf.split(X):
            reg_y = LGBMRegressor(n_estimators=300, random_state=random_state+10)
            reg_y.fit(np.hstack([X[tr], T_bin[tr].reshape(-1,1), M[tr].reshape(-1,1)]), Y[tr])
            X_te = X[te]
            y_1_m1[te] = reg_y.predict(np.hstack([X_te, np.ones((len(te),1)), m_hat_1[te].reshape(-1,1)]))
            y_1_m0[te] = reg_y.predict(np.hstack([X_te, np.ones((len(te),1)), m_hat_0[te].reshape(-1,1)]))
            y_0_m0[te] = reg_y.predict(np.hstack([X_te, np.zeros((len(te),1)), m_hat_0[te].reshape(-1,1)]))
        y_1_m1[:] = np.clip(y_1_m1, 0, 1); y_1_m0[:] = np.clip(y_1_m0, 0, 1); y_0_m0[:] = np.clip(y_0_m0, 0, 1)
        te = float(np.mean(y_1_m1 - y_0_m0)); nde = float(np.mean(y_1_m0 - y_0_m0)); nie = te-nde
        delta_m = float(np.mean(m_hat_1 - m_hat_0))
        return {'TE':te,'NDE':nde,'NIE':nie,'Delta_M':delta_m}

    def exp_residual():
        m_hat_0 = np.zeros(n); m_hat_1 = np.zeros(n)
        lin = LinearRegression()
        for tr, te in kf.split(X):
            lin.fit(np.hstack([X[tr], T_bin[tr].reshape(-1,1)]), M[tr])
            m_hat_0[te] = lin.predict(np.hstack([X[te], np.zeros((len(te),1))]))
            m_hat_1[te] = lin.predict(np.hstack([X[te], np.ones((len(te),1))]))
        lin_full = LinearRegression().fit(np.hstack([X, T_bin.reshape(-1,1)]), M)
        m_pred_full = lin_full.predict(np.hstack([X, T_bin.reshape(-1,1)]))
        resid = M - m_pred_full
        m_cf0 = m_hat_0 + resid
        m_cf1 = m_hat_1 + resid
        y_1_m1 = np.zeros(n); y_1_m0 = np.zeros(n); y_0_m0 = np.zeros(n)
        for tr, te in kf.split(X):
            reg_y = LGBMRegressor(n_estimators=300, random_state=random_state+10)
            reg_y.fit(np.hstack([X[tr], T_bin[tr].reshape(-1,1), M[tr].reshape(-1,1)]), Y[tr])
            X_te = X[te]
            y_1_m1[te] = reg_y.predict(np.hstack([X_te, np.ones((len(te),1)), m_cf1[te].reshape(-1,1)]))
            y_1_m0[te] = reg_y.predict(np.hstack([X_te, np.ones((len(te),1)), m_cf0[te].reshape(-1,1)]))
            y_0_m0[te] = reg_y.predict(np.hstack([X_te, np.zeros((len(te),1)), m_cf0[te].reshape(-1,1)]))
        y_1_m1[:] = np.clip(y_1_m1,0,1); y_1_m0[:] = np.clip(y_1_m0,0,1); y_0_m0[:] = np.clip(y_0_m0,0,1)
        te = float(np.mean(y_1_m1 - y_0_m0)); nde = float(np.mean(y_1_m0 - y_0_m0)); nie = te-nde
        delta_m = float(np.mean(m_hat_1 - m_hat_0))
        return {'TE':te,'NDE':nde,'NIE':nie,'Delta_M':delta_m}

    def exp_contT():
        m_hat_0 = np.zeros(n); m_hat_1 = np.zeros(n)
        T_cont = T_raw.astype(float)
        for tr, te in kf.split(X):
            reg_m = LGBMRegressor(n_estimators=200, random_state=random_state)
            reg_m.fit(np.hstack([X[tr], T_cont[tr].reshape(-1,1)]), M[tr])
            m_hat_0[te] = reg_m.predict(np.hstack([X[te], np.full((len(te),1), t0)]))
            m_hat_1[te] = reg_m.predict(np.hstack([X[te], np.full((len(te),1), t1)]))
        y_1_m1 = np.zeros(n); y_1_m0 = np.zeros(n); y_0_m0 = np.zeros(n)
        for tr, te in kf.split(X):
            reg_y = LGBMRegressor(n_estimators=300, random_state=random_state+10)
            reg_y.fit(np.hstack([X[tr], T_cont[tr].reshape(-1,1), M[tr].reshape(-1,1)]), Y[tr])
            X_te = X[te]
            y_1_m1[te] = reg_y.predict(np.hstack([X_te, np.full((len(te),1), t1), m_hat_1[te].reshape(-1,1)]))
            y_1_m0[te] = reg_y.predict(np.hstack([X_te, np.full((len(te),1), t1), m_hat_0[te].reshape(-1,1)]))
            y_0_m0[te] = reg_y.predict(np.hstack([X_te, np.full((len(te),1), t0), m_hat_0[te].reshape(-1,1)]))
        y_1_m1[:] = np.clip(y_1_m1,0,1); y_1_m0[:] = np.clip(y_1_m0,0,1); y_0_m0[:] = np.clip(y_0_m0,0,1)
        te = float(np.mean(y_1_m1 - y_0_m0)); nde = float(np.mean(y_1_m0 - y_0_m0)); nie = te-nde
        delta_m = float(np.mean(m_hat_1 - m_hat_0))
        return {'TE':te,'NDE':nde,'NIE':nie,'Delta_M':delta_m}

    return {
        'baseline': baseline_reg_clip(),
        'scaled': exp_scaled(),
        'residual': exp_residual(),
        'contT': exp_contT()
    }

# Run
res_a = run_experiments_on_df(df_a, conf_cols)
res_b = run_experiments_on_df(df_b, conf_cols)

# If get_ground_truth available:
try:
    gt_a = get_ground_truth(df_a, PARAMS_A)
    gt_b = get_ground_truth(df_b, PARAMS_B)
except Exception:
    gt_a = (np.nan, np.nan, np.nan)
    gt_b = (np.nan, np.nan, np.nan)

def pretty_print(name, gt, res):
    print(f"\n{'='*60}\n{name}\n{'='*60}")
    print(f"Ground Truth: TE={gt[0]:.3f}, NDE={gt[1]:.3f}, NIE={gt[2]:.3f}\n")
    for k,v in res.items():
        print(f"[{k}] TE={v['TE']:.3f}, NDE={v['NDE']:.3f}, NIE={v['NIE']:.3f}, ΔM={v['Delta_M']:.3f}")
pretty_print("Dataset A", gt_a, res_a)
pretty_print("Dataset B", gt_b, res_b)

# Return results DataFrame for further analysis
rows=[]
for ds, res, gt in [('A',res_a,gt_a),('B',res_b,gt_b)]:
    for k,v in res.items():
        rows.append({'dataset':ds,'exp':k,'TE':v['TE'],'NDE':v['NDE'],'NIE':v['NIE'],
                     'Delta_M':v['Delta_M'],'GT_TE':gt[0],'GT_NDE':gt[1],'GT_NIE':gt[2]})
pd.DataFrame(rows)
```
